# LSM - Nahuatl

## Que vamos a hacer?

Traducir de LSM a Nahuatl hablado en tiempo real


## Como lo vamos a hacer?


Usando las herramientas de Mediapipe Hands, Entrenaremos una red neuronal que sea capaz de trancribir o traducir el lenguaje de senas mexicano en palabras en nahuatl


Mediapipe Hands es un modelo de vision por computadora, detecta en tiempo real 21 puntos clave 3D (landmarls) de la mano (articulaciones, puntas de dedos, palma) en imágenes o video. Permite identificar manos izquierdas/derechas y rastrear movimientos precisos.
